```python
import pandas as pd
from ydata_profiling import ProfileReport

# --- 1. CONFIGURATION ---
csv_file = '../data/StockEtablissement_utf8_100000.csv'
output_html = 'rapport_exploration_sirene.html'

print("⏳ Chargement des données...")
# On lit tout en 'str' (texte) pour éviter les erreurs de lecture, puis on convertit après
df = pd.read_csv(csv_file, dtype=str)

# --- 2. CONVERSION DES DATES (Crucial pour ton analyse) ---
# On convertit les colonnes dates pour que le rapport détecte les trous temporels
print("⚙️ Conversion des dates...")
df['dateCreationEtablissement'] = pd.to_datetime(df['dateCreationEtablissement'], errors='coerce')
df['dateDernierTraitementEtablissement'] = pd.to_datetime(df['dateDernierTraitementEtablissement'], errors='coerce')

# --- 3. GÉNÉRATION DU RAPPORT ---
print(f"📊 Génération du rapport sur {len(df)} lignes... (Cela peut prendre 1 à 2 minutes)")

# 'minimal=True' désactive les corrélations lourdes pour aller vite et ne pas faire planter Docker
profile = ProfileReport(df, title="Audit Qualité - Stock Etablissement", minimal=True)

# Sauvegarde du fichier HTML (pour ton rendu)
profile.to_file(output_html)

print(f"\n✅ SUCCÈS ! Le rapport est sauvegardé sous : {output_html}")
print("👉 Regarde dans la colonne de gauche de Jupyter pour le télécharger.")

# Affichage direct dans le notebook
profile.to_notebook_iframe()
```
